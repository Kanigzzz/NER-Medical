```python
from datasets import load_dataset
from transformers import AutoTokenizer
import pandas as pd
```

```python
dataset = load_dataset("bigbio/bc5cdr", name="bc5cdr_bigbio_kb", trust_remote_code=True)
```

```python
MODEL_NAME = "microsoft/BiomedNLP-BiomedBERT-base-uncased-abstract-fulltext"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

test_text = "Naloxone reverses the antihypertensive effect of clonidine."

encoding = tokenizer(
    test_text,
    return_offsets_mapping=True,
    add_special_tokens=True
)

for token_id, offset in zip(encoding['input_ids'], encoding['offset_mapping']):
    token = tokenizer.convert_ids_to_tokens(token_id)
    print(f"token: {token:20} | chars: {offset}")
```
